```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model = "gpt-5")

text = "Hello, My name is Manish Sachan, " \
"Email is Manishsachan069@gmail.com and my age is 29"

res = llm.invoke(f"Please give me only name, email and age from this text: {text}")
print(res.content)
```

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStructure(BaseModel):
    name:str = Field(description="Complete Name")
    email:str = Field(description="Email Address") 
    age:int = Field(description="Age")

structured_llm=llm.with_structured_output(ResponseStructure)
res=structured_llm.invoke(f"Please give me only name, email and age from this text: {text}")

print(res.model_dump())

```

```python
class Movies(BaseModel):
    title:str= Field(description="Name of the movie"),
    year:int= Field(description="Year of release")

class All_movies(BaseModel):
    movies:List[Movies]

show_movies=llm.with_structured_output(All_movies)
res=show_movies.invoke("give me result of top three comedy movies")
res
```
