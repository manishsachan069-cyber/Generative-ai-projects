```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm=ChatOpenAI(model="gpt-4")
promts=ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator and translate input into {language}"},
    {"role":"user", "content":"{query}"}
])

chains= promts | llm
res=chains.invoke({"language":"Hinglish", "query":"I am a GenAI developer and I am earning INR 2lakhs per month"})
print(res.content)
```
